```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

import data file

```python
df = pd.read_csv('/Spotify-2000 (2).csv')
```

```python
df.head()
```

data information

```python
df.info()
```

Line Graph(Popularity vs Year)

```python
import matplotlib.pyplot as plt

plt.bar(df['Year'], df['Popularity'])

plt.xlabel('Year')
plt.ylabel('Popularity')
plt.title('Popular song relaes year')

plt.show()
```

```python
df.head()
```

Title vs Popularity

```python
plt.barh(df['Title'], df['Popularity'])
plt.title('popular song')

plt.xlabel('Title')
plt.ylabel('Popularity')
plt.show()
```

```python
df.head()
```

Title vs Artist


```python
plt.scatter(df['Title'], df['Artist'])
plt.xlabel('Title')
plt.ylabel('Artist')
plt.title('Title vs Artist')

plt.show()
```

Artist vs Year

```python
plt.scatter(df['Artist'], df['Year'])

plt.xlabel('Artist')
plt.ylabel('Year')
plt.title('Artist vs Year')

plt.show()
```
